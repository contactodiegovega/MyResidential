# Creación de la startup

## 1. Ideas iniciales

### Idea 1 — Gestión inteligente de comunidades de propietarios

Plataforma para facilitar la gestión de comunidades, centralizando información sobre gastos, proveedores, incidencias y comunicaciones.

Los datos permitirían analizar gastos, comparar proveedores, detectar desviaciones y mejorar la transparencia y la toma de decisiones.

---

### Idea 2 — Previsión de recursos en carreras populares

Herramienta para organizadores de carreras que utilice datos históricos de participantes, recorrido, meteorología y ediciones anteriores para estimar las necesidades de recursos.

El objetivo sería mejorar la planificación de elementos como agua, avituallamiento, personal y material necesario para cada evento.

---

### Idea 3 — Gestión inteligente de stock y desperdicio alimentario

Herramienta dirigida a supermercados o grupos de tiendas de alimentación para mejorar la gestión de productos perecederos.

Los datos de ventas, inventario, fechas de caducidad, productos y pedidos permitirían detectar productos con riesgo de desperdicio o falta de existencias y ayudar a planificar mejor el stock.

---

# 2. Idea elegida

## Gestión inteligente de comunidades de propietarios

### Nombre

**MyResidential**

### Sector

**PropTech / Gestión de comunidades residenciales**

### Qué hace

Plataforma web que centraliza y analiza los datos de comunidades de propietarios para facilitar su gestión y mejorar la transparencia y la toma de decisiones.

### Cliente objetivo

**Administradores de fincas y comunidades de propietarios.**

### Problema que resuelve

La información sobre gastos, proveedores, incidencias y gestión de una comunidad puede encontrarse dispersa y resultar difícil de analizar y comparar.

### Propuesta de valor

Convertir los datos de las comunidades en información útil para mejorar su gestión, detectar desviaciones, comparar proveedores y facilitar decisiones más transparentes.

### Producto principal

Plataforma de gestión y análisis de comunidades con:

* Seguimiento de gastos.
* Transparencia en los contratos.
* Gestión de proveedores.
* Seguimiento de incidencias.
* Comparativas entre comunidades similares.
* Visualización de indicadores clave.
* Gestión de reservas de espacios comunes
* Chat intervecinal.
* Tablón de anuncios.
