# Creación de la startup

##  Ideas iniciales

### Idea 1 — Gestión inteligente de comunidades de propietarios

Plataforma para facilitar la gestión de comunidades, centralizando información sobre gastos, proveedores, incidencias y comunicaciones.

Los datos permitirían analizar gastos, comparar proveedores, detectar desviaciones y mejorar la transparencia y la toma de decisiones.

---

### Idea 2 — Previsión de recursos en carreras populares

Herramienta para organizadores de carreras que utilice datos históricos de participantes, recorrido, meteorología y ediciones anteriores para estimar las necesidades de recursos.

El objetivo sería mejorar la planificación de elementos como agua, avituallamiento, personal y material necesario para cada evento.

---

### Idea 3 — LocalFinder: localización inteligente de nuevos establecimientos

Herramienta dirigida a cadenas y franquicias que quieran abrir nuevos establecimientos.

Los datos de población, renta, competencia, transporte, negocios cercanos y características de cada zona permitirían identificar y comparar ubicaciones con potencial para nuevas aperturas.

#  Idea seleccionada

## MyResidential

### Sector
PropTech / Gestión de comunidades residenciales.

### ¿Qué hace?
MyResidential es una plataforma para administradores de fincas que centraliza la gestión de sus comunidades y utiliza los datos para facilitar la toma de decisiones.

La plataforma cuenta con dos áreas principales:

- **Panel del administrador:** permite gestionar varias comunidades, analizar gastos, proveedores e incidencias y comparar comunidades con características similares.
- **Panel del vecino:** permite consultar gastos, contratos, proveedores y documentación de su comunidad, además de acceder al tablón de anuncios y otras funciones de gestión.

### Cliente objetivo
El cliente principal son los **administradores de fincas que gestionan múltiples comunidades de propietarios**.

Los vecinos y propietarios serían usuarios de la plataforma, pero no el cliente principal.

### Problema
Los administradores gestionan una gran cantidad de información procedente de distintas comunidades: gastos, contratos, proveedores, incidencias y documentación.

Aunque disponen de esta información, puede resultar difícil analizarla de forma conjunta y comparar el funcionamiento de diferentes comunidades.

Además, los propietarios no siempre disponen de una forma sencilla y centralizada de consultar la información relacionada con su comunidad.

### Propuesta de valor
MyResidential transforma los datos generados por las comunidades en información útil para administradores y propietarios.

Permite centralizar la gestión, aumentar la transparencia y comparar comunidades similares para detectar desviaciones, diferencias de costes y posibles oportunidades de mejora.

### Producto principal
La plataforma contará con dos paneles:

#### Panel del administrador
- Gestión de múltiples comunidades.
- Análisis de gastos.
- Gestión y comparación de proveedores.
- Seguimiento de incidencias.
- Comparación entre comunidades similares.
- Indicadores y visualizaciones de gestión.

#### Panel del vecino
- Consulta de gastos de la comunidad.
- Consulta de contratos y proveedores.
- Documentación de la comunidad.
- Seguimiento y comunicación de incidencias.
- Tablón de anuncios.
- Reserva de espacios comunes.

#  Cliente ficticio

## Urbalia Gestión de Fincas

Urbalia Gestión de Fincas es una empresa ficticia dedicada a la administración de comunidades de propietarios en Madrid.

Actualmente gestiona aproximadamente **40 comunidades**, que representan unas **2.500 viviendas**.

Las comunidades presentan características diferentes: algunas cuentan con piscina, garaje, ascensores, zonas verdes, conserjería o instalaciones deportivas, mientras que otras disponen únicamente de servicios básicos.

La empresa trabaja con numerosos proveedores externos para prestar servicios como:

- Limpieza.
- Mantenimiento de ascensores.
- Jardinería.
- Piscinas.
- Seguridad.
- Fontanería.
- Electricidad.
- Mantenimiento de garajes.
- Reparaciones generales.

Actualmente, gran parte de la información de cada comunidad se consulta de forma independiente, lo que dificulta obtener una visión conjunta de todas las comunidades gestionadas.

## Problema del cliente

Urbalia dispone de una gran cantidad de información sobre sus comunidades, pero tiene dificultades para analizarla de forma global.

Por ejemplo, puede conocer cuánto paga una comunidad por mantenimiento de ascensores, pero no resulta sencillo saber si ese coste es elevado o reducido respecto a otras comunidades similares.

Lo mismo ocurre con proveedores, incidencias, tiempos de resolución y otros gastos.

Además, los propietarios demandan cada vez mayor acceso y transparencia sobre la información relacionada con su comunidad.

## Qué necesita Urbalia

Urbalia necesita una plataforma que permita:

- Centralizar la información de todas sus comunidades.
- Consultar gastos y contratos.
- Gestionar proveedores.
- Registrar y hacer seguimiento de incidencias.
- Comparar comunidades con características similares.
- Detectar desviaciones o comportamientos poco habituales.
- Facilitar información a los propietarios mediante un panel específico.

MyResidential busca cubrir estas necesidades mediante una plataforma de gestión y análisis de datos.
